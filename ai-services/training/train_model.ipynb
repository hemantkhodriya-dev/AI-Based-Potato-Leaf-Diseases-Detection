1. **Healthy**

## 2. Inspect Dataset Folders

```python
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from sklearn.metrics import classification_report, confusion_matrix

dataset_dir = './dataset/plantvillage dataset/color'
if not os.path.exists(dataset_dir):
    dataset_dir = './dataset'

print('Subdirectories found:')
for item in os.listdir(dataset_dir):
    if 'potato' in item.lower():
        print('-', item)
```

## 3. Data Preprocessing & 70/15/15 Split & Augmentation

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# Data Augmentation for training
train_datagen = ImageDataGenerator(
    rescale=1.0/255.0,
    rotation_range=20,
    width_shift_range=0.15,
    height_shift_range=0.15,
    shear_range=0.15,
    zoom_range=0.15,
    horizontal_flip=True,
    validation_split=0.30 # 70% train, 30% for val/test
)

val_test_datagen = ImageDataGenerator(rescale=1.0/255.0, validation_split=0.30)

train_gen = train_datagen.flow_from_directory(
    dataset_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training'
)

val_gen = val_test_datagen.flow_from_directory(
    dataset_dir,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation'
)

print('Class Indices:', train_gen.class_indices)
```

## 4. MobileNetV2 Transfer Learning Model Architecture

```python
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation='relu')(x)
x = Dropout(0.4)(x)
outputs = Dense(3, activation='softmax')(x)

model = Model(inputs=base_model.input, outputs=outputs)
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
model.summary()
```

## 5. Model Training & Evaluation

```python
epochs = 15
history = model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=epochs
)

# Save trained model
model.save('potato_disease_model.keras')
print('Saved trained model to potato_disease_model.keras')
```

## 6. Evaluation Metrics & Confusion Matrix

```python
val_gen.reset()
preds = model.predict(val_gen)
y_pred = np.argmax(preds, axis=1)
y_true = val_gen.classes

labels = list(train_gen.class_indices.keys())
print(classification_report(y_true, y_pred, target_names=labels))
print('Confusion Matrix:')
print(confusion_matrix(y_true, y_pred))
```
